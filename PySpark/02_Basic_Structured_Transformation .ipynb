```python
# Small Example for schema 

schema_string = "name string, age int"

from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema_spark = StructType([
StructField("name", StringType(), True),
StructField("age", IntegerType(), True)
])
```

```python
#Columnds and Expressions 

from pyspark.sql.functions import col, expr
emp.salary
```

```python
#Select columns 
#Select employee_id, name, age, salary from emp 

emp_filtered = emp.select(col("employee_id"), emp['name'], expr("age"), emp.salary)
```

```python
#Show filtered dataFrame (Action)
emp_filtered.show()
```

```python
#using expr for select 
#select employee_id as emp_id, name, cast(age as int) as age, salary from emp_filtered 

emp_casted = emp_filtered.select(expr('employee_id as emp_id'), col('name'), expr('cast(age as int) as age'), col('salary'))
```

```python
#Show dataFrame (Action)
emp_casted.show()
print("__________________\nSchema of emp_casted")
emp_casted.printSchema()
```

```python
#Another way of filtering data without too many syntax 
#select employee_id as emp_id, name, cast(age as int) as age, salary from emp_filtered 

emp_casted1 = emp_filtered.selectExpr("employee_id as emp_id", "name", "cast(age as int) as age", "salary")
```

```python
#Show dataFrame (Action)
emp_casted1.show()
print("__________________\nSchema of emp_casted")
emp_casted1.printSchema()
```

```python
#Filtern emp based on age > 30
#Select emp_id, name, age, salary from employee_casted where age > 30

emp_final = emp_casted.select("emp_id", "name", "age", "salary").where("age > 30")
```

```python
#Show dataFrame (Action)

emp_final.show()
```

```python
#Write the data back as csv 
emp_final.write.format("csv").save("data/output/2/emp.csv")
```
